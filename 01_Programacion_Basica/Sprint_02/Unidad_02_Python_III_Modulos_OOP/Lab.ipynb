```python
# Implementa aquí la clase Perro
class Perro:
    """Representa un perro con sus características físicas y su estado.

    Atributos de clase:
        num_patas (int): Número de patas del perro (4).
        num_orejas (int): Número de orejas del perro (2).
        num_ojos (int): Número de ojos del perro (2).

    Atributos de instancia:
        raza (str): Raza del perro.
        pelo (str): Color del pelo. Por defecto "Marrón".
        duenio (str | None): Dueño del perro. Por defecto None (sin dueño).
        velocidad (int | float): Velocidad actual en km/h. Por defecto 0.
    """
    
    num_patas = 4
    num_orejas = 2
    num_ojos = 2

    def __init__(self, raza, pelo = "Marrón", duenio = None, velocidad = 0):
        self.raza = raza
        self.pelo = pelo
        self.duenio = duenio
        self.velocidad = velocidad

    def andar(self, aumento_velocidad):
        self.velocidad += aumento_velocidad
        return f"Mi perro va a {self.velocidad} km/h."

    def parar(self): 
        self.velocidad = 0
        return "Mi perro para"

    def ladrar(self, senal):
        if senal == "Ladra":
            return "GUAU"
        else:
            return "No pasa nada"

#Crea un objeto de tipo Perro, sin dueño, en mi_perro.
mi_perro = Perro("Teckel")

#Comprueba que están bien todos sus atributos
print(mi_perro.num_patas, mi_perro.num_orejas, mi_perro.num_ojos, mi_perro.raza, mi_perro.pelo, mi_perro.duenio, mi_perro.velocidad)

#Prueba que ande, y comprueba su velocidad.
mi_perro.andar(30)

#Páralo.
print(mi_perro.parar())

#Ladra
print(mi_perro.ladrar("Ladra"))
```

```python
#1.Implementar clase Tienda
class Tienda:
    """
    Atributos comunes a todas las tiendas. Tipo que vale "Electrodomésticos", y abierta que vale True
    Atributos propios de cada tienda: nombre, direccion, num_empleados, ventas de los últimos 3 meses.
    """ 
    tipo = "Electrodomésticos"
    abierta = True

    def __init__(self, nombre, direccion, num_empleados, ventas_3_meses):
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_3_meses = ventas_3_meses

    def total_ventas(self):
        """Suma del total de ventas de las tiendas"""
        return sum(self.ventas_3_meses)

    def ventas_por_empleado(self):
        """Método que calcula la media de ventas de los últimos meses, por empleado"""
        return self.total_ventas() / self.num_empleados

    def nombre_direccion(self):
        """Método que devuelve en un string el nombre de la tienda, junto con su dirección"""
        return f"{self.nombre} : {self.direccion}"

    def ventas_ultimo_mes(self):
        """Método que obtiene las ventas del último mes"""
        return self.ventas_3_meses[-1]

    def proyeccion_marketing(self, x):
        """Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda,\
        siendo X el parámetro de entrada"""
        if x == 0:
            factor = 1
        
        elif x < 1000:
            factor = 1.2
        else:
            factor = 1.5

        self.ventas_3_meses = [venta * factor for venta in self.ventas_3_meses]
        return self.ventas_3_meses

#2. Crear 3 tiendas
tienda1 = Tienda("Electro Sol", "Avenida de la Constitución 12, Sevilla", 8, [12000, 15000, 13500])
tienda2 = Tienda("Hogar Plus", "Calle Mayor 5, Madrid", 5, [8000, 9500, 10000])
tienda3 = Tienda("TecnoHogar", "Avenida Diagonal 45, Barcelona", 12, [20000, 18000, 22000])

#3. Comprobar en al menos una de ellas todo lo implementado en la clase (sus atributos, media de ventas, ventas por empleado...).
print("Tipo:", tienda1.tipo)                                  
print("Abierta:", tienda1.abierta)                            
print("Nombre:", tienda1.nombre)
print("Dirección:", tienda1.direccion)
print("Empleados:", tienda1.num_empleados)
print("Ventas últimos 3 meses:", tienda1.ventas_3_meses)
print("Total ventas:", tienda1.total_ventas())                            
print("Ventas por empleado:", tienda1.ventas_por_empleado())  
print("Nombre y dirección:", tienda1.nombre_direccion())
print("Ventas último mes:", tienda1.ventas_ultimo_mes())      
print("Proyección (500 €):", tienda1.proyeccion_marketing(500))   
print("Proyección (2000 €):", tienda1.proyeccion_marketing(2000)) 

# 4. Suma de las ventas del último mes de todas las tiendas
total_ultimo_mes = 0
for tienda in [tienda1, tienda2, tienda3]:
    total_ultimo_mes += tienda.ventas_ultimo_mes()
print("Total último mes:", total_ultimo_mes)                 

# 5. Tiendas cuya dirección contiene "Avenida"
for tienda in [tienda1, tienda2, tienda3]:
    if "Avenida" in tienda.direccion:
        print(tienda.nombre)  

    

```
