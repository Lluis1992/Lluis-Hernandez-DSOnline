```python
import math

def dot_product(vec1, vec2, precision = 2):
    producto = 0
    for indice in range(len(vec1)): # Importante sólo podemos hacer producto escalar de vectores con el mismo número de componentes o dimensiones
        producto += vec1[indice] * vec2[indice]
    producto = round(producto, precision)
    return producto

def norm_vec(vec, precision = 1):
    escalar = dot_product(vec,vec) # Esta es la magia de las funciones y los métodos construir unos sobre otros :-), reusar, craft, craft
    resultado = math.sqrt(escalar)
    resultado = round(resultado, precision)
    return resultado

def cos_vec(vec1, vec2, precision = 1):
    resultado =dot_product(vec1,vec2)/(norm_vec(vec1) * norm_vec(vec2))
    resultado = round(resultado, precision)
    return resultado

# vec1, ya sin su quinta componente
vec1 = [12, -4, 3, 0]
vec2 = [1, 3, -0.33, -12.1]
vec3 = [-2, 0, -1, 0]
vec4 = [0, 0, 1, 1]
nombres = ["vec1", "vec2", "vec3", "vec4"]
lista_vectores = [vec1, vec2, vec3, vec4]
cosenos = []
for i in range(len(lista_vectores)):
    for j in range(i + 1, len(lista_vectores)):
        cosenos.append(cos_vec(lista_vectores[i], lista_vectores[j]))
print(cosenos)
```
